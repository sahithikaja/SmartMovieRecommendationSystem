```python
def personalized_recommend(movie, language="Any", min_rating=7.0, max_runtime=180):
    index = movies[movies['title'] == movie].index[0]
    distances = list(enumerate(similarity[index]))
    distances = sorted(distances, key=lambda x: x[1], reverse=True)
    count = 0
    print(f"\nPersonalized recommendations for '{movie}':\n")
    for i in distances[1:]:
        row = movies.iloc[i[0]]
        if language != "Any" and row['original_language'] != language:
            continue
        if row['vote_average'] < min_rating:
            continue
        if row['runtime'] > max_runtime:
            continue
        print(f"🎬 {row['title']}")
        print(f"⭐ Rating: {row['vote_average']}")
        print(f"⏱ Runtime: {int(row['runtime'])} min")
        print(f"🌐 Language: {row['original_language']}")
        print("Why recommended?")
        print("✓ Similar to your selected movie")
        print("✓ Matches your preferences")
        print("-"*35)
        count += 1
        if count == 5:
            break
```

```python
personalized_recommend(
    "Interstellar",
    language="en",
    min_rating=8,
    max_runtime=150
)
```

```python
import requests

movie_id = data["results"][0]["id"]

url = f"https://api.themoviedb.org/3/movie/{movie_id}"

params = {
    "append_to_response": "credits,keywords"
}

response = requests.get(url, headers=headers, params=params)

movie_details = response.json()

print("Title:", movie_details["title"])
print("Overview:", movie_details["overview"])
print("Genres:", [g["name"] for g in movie_details["genres"]])
print("Language:", movie_details["original_language"])
```

```python
def create_tmdb_tags(movie):
    parts = []
    parts.append(movie.get("overview", ""))
    for genre in movie.get("genres", []):
        parts.append(genre.get("name", ""))
    keywords = movie.get("keywords", {}).get("keywords", [])
    for keyword in keywords:
        parts.append(keyword.get("name", ""))
    cast = movie.get("credits", {}).get("cast", [])
    for person in cast[:5]:
        parts.append(person.get("name", ""))
    crew = movie.get("credits", {}).get("crew", [])
    for person in crew:
        if person.get("job") == "Director":
            parts.append(person.get("name", ""))
    return " ".join(parts).lower()
tmdb_tags = create_tmdb_tags(movie_details)
print("Movie:", movie_details["title"])
print("Tags:")
print(tmdb_tags)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
tmdb_vector = tfidf.transform([tmdb_tags])
scores = cosine_similarity(tmdb_vector, vectors)[0]
top_indexes = scores.argsort()[::-1][:10]
print("Movies similar to:", movie_details["title"])
print()
for i in top_indexes:
    print(movies.iloc[i]["title"], "->", round(scores[i], 3))
```

```python
top_indexes = scores.argsort()[::-1]
recommendations = []
for i in top_indexes:
    title = movies.iloc[i]["title"]
    if title.lower() == movie_details["title"].lower():
        continue
    recommendations.append((title, scores[i]))
    if len(recommendations) == 5:
        break
print("Top 5 recommendations:")
print()
for title, score in recommendations:
    print(title, "->", round(score, 3))
```

```python
url = f"https://api.themoviedb.org/3/movie/{movie_id}/similar"
params = {
    "language": "en-US",
    "page": 1
}
response = requests.get(url, headers=headers, params=params)
similar_data = response.json()
print("TMDB similar movies:")
print()
for movie in similar_data["results"][:10]:
    print(movie["title"])
```

```python
tmdb_candidates = []

for movie in similar_data["results"][:10]:

    tags = (
        movie.get("title", "") + " " +
        movie.get("overview", "") + " " +
        " ".join(str(x) for x in movie.get("genre_ids", []))
    ).lower()

    tmdb_candidates.append({
        "id": movie["id"],
        "title": movie["title"],
        "tags": tags,
        "rating": movie.get("vote_average", 0),
        "language": movie.get("original_language", "")
    })

print("TMDB candidates:", len(tmdb_candidates))

for movie in tmdb_candidates:
    print(movie["title"])
```

```python
tmdb_scores = []

for movie in tmdb_candidates:

    movie_vector = tfidf.transform([movie["tags"]])

    score = cosine_similarity(tmdb_vector, movie_vector)[0][0]

    tmdb_scores.append({
        "title": movie["title"],
        "score": score,
        "rating": movie["rating"],
        "language": movie["language"],
        "id": movie["id"]
    })

tmdb_scores = sorted(
    tmdb_scores,
    key=lambda x: x["score"],
    reverse=True
)

print("TMDB movies ranked by your TF-IDF model:")
print()

for movie in tmdb_scores:
    print(movie["title"], "->", round(movie["score"], 3))
```

```python
# Local recommendations
local_results = []
for title, score in recommendations:
    row = movies[movies["title"] == title].iloc[0]

    local_results.append({
        "title": title,
        "score": score,
        "source": "Your Dataset",
        "rating": row["vote_average"],
        "language": row["original_language"]
    })
external_results = []
for movie in tmdb_scores:
    external_results.append({
        "title": movie["title"],
        "score": movie["score"],
        "source": "TMDB",
        "rating": movie["rating"],
        "language": movie["language"]
    })
# Combine both
combined_results = local_results + external_results
combined_results = sorted(
    combined_results,
    key=lambda x: x["score"],
    reverse=True
)
top_5 = combined_results[:5]
print("Top 5 Recommendations:\n")
for i, movie in enumerate(top_5, 1):
    print(
        i,
        movie["title"],
        "->",
        round(movie["score"], 3),
        "->",
        movie["source"]
    )
```

```python
# Give rank scores to local movies
for i, movie in enumerate(local_results):
    movie["rank_score"] = 1 - (i / len(local_results))
for i, movie in enumerate(external_results):
    movie["rank_score"] = 1 - (i / len(external_results))
combined_results = local_results + external_results
combined_results = sorted(
    combined_results,
    key=lambda x: x["rank_score"],
    reverse=True
)
top_5 = combined_results[:5]
print("Top 5 Recommendations:\n")
for i, movie in enumerate(top_5, 1):
    print(
        i,
        movie["title"],
        "->",
        round(movie["rank_score"], 3),
        "->",
        movie["source"]
    )
```
