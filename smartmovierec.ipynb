```python
movies = movies[
    ['id',
     'title',
     'genres',
     'keywords',
     'overview',
     'runtime',
     'original_language',
     'vote_average',
     'vote_count']
]
```

```python
credits = credits[
    ['movie_id',
     'title',
     'cast',
     'crew']
]
```

```python
movies['overview'] = movies['overview'].fillna('')
```

```python
movies['runtime'] = movies['runtime'].fillna(movies['runtime'].median())
```

```python
movies.isnull().sum()
```

```python
credits.isnull().sum()
```

```python
movies = movies.merge(
    credits,
    left_on='id',
    right_on='movie_id'
)
```

```python
movies.columns
```

```python
movies.drop(['title_y', 'movie_id'], axis=1, inplace=True)
```

```python
movies.rename(columns={'title_x': 'title'}, inplace=True)
```

```python
movies.columns
```

```python
movies.to_csv("smartmovie_dataset.csv", index=False)
```

```python
movies[['title', 'genres', 'keywords', 'cast', 'crew']].head()
```

```python
def convert_names(text):
    result = []
    
    for item in ast.literal_eval(text):
        result.append(item['name'])
        
    return result
```

```python
movies['genres'] = movies['genres'].apply(convert_names)
movies['keywords'] = movies['keywords'].apply(convert_names)
```

```python
movies[['title', 'genres', 'keywords']].head()
```

```python
def convert_cast(text):
    result = []
    
    for item in ast.literal_eval(text)[:3]:
        result.append(item['name'])
        
    return result
```

```python
movies['cast'] = movies['cast'].apply(convert_cast)
```

```python
movies[['title', 'cast']].head()
```

```python
def get_director(text):
    for item in ast.literal_eval(text):
        if item['job'] == 'Director':
            return item['name']
    
    return ''
```

```python
movies['director'] = movies['crew'].apply(get_director)
```

```python
movies[['title', 'director']].head()
```

```python
movies.drop('crew', axis=1, inplace=True)
```

```python
movies['overview'] = movies['overview'].apply(lambda x: x.split())
```

```python
movies['genres'] = movies['genres'].apply(
    lambda x: [i.replace(" ", "") for i in x]
)
movies['keywords'] = movies['keywords'].apply(
    lambda x: [i.replace(" ", "") for i in x]
)
movies['cast'] = movies['cast'].apply(
    lambda x: [i.replace(" ", "") for i in x]
)
movies['director'] = movies['director'].apply(
    lambda x: x.replace(" ", "")
)
```

```python
movies['tags'] = (
    movies['overview']
    + movies['genres']
    + movies['keywords']
    + movies['cast']
    + movies['director'].apply(lambda x: [x])
)
```
